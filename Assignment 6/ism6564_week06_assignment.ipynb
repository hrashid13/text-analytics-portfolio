### Your answer — Task 1, the brief

In 150–250 words, written by you: the organisation and the job (what text arrives, what object must come out, who or what acts on it); the constraints that make a **small, local, fine-tuned** model the right shape (privacy, hardware, volume, cost) rather than a prompt on a rented model or retrieval; what *good* means (which field matters most and why); and one thing the model must **not** be trusted to know, because it is a fact that changes (that is Week 7).

Organization: Pharmacy

Input: Typed form of a refill request from a fax/email

Privacy Constraint: Name, DOB, Medication all fall under HIPAA

Why Small and Local Model: Outside of the information falling under HIPAA, this is not a taxing job, just extracting text

Field that matters the most: Patient + Medication, don't want to give wrong medications to patients

Don't trust item: The patient's profile, the model does not touch the patients actual profile with all their medication, just extracts the text from the request

Who consumes output: The pharmacist, so a licensed professional is always making the final decision, a patient can send a request to get their Xanax filled a week early, and the model would still extract the request, but the pharmacist would deny it because it is too early (state law)

Why a form is not enough: This is a free text form, so a form can't be imposed

Fixed choice field: callback_reason: when a patient's request has insufficient data the pharmacy needs to call back, callback_reason can be, missing_information, consultation_needed, invalid_refill_amount, invalid_dob and none (if everything is correct)

Rules for this:

if DOB is invalid (month 13, February 30th) then callback

if medication field is empty then call back

if name field empty then call back

if consultation is needed due to side effects stated then call back

if refill amount exceeds 90 days (or 30 for controlled medications) then call back


````python
# --- 1a: the contract and the minimal prompt -----------------------------------------------
# Record: at least three fields; at least one Literal (a fixed choice) and at least one non-Literal
# (StrictBool, int/float, or a str with a pattern). Name them exactly as your data columns will be.
class Record(BaseModel):
    patient_name: str
    date_of_birth: str
    medication: str
    refill_days: int
    is_controlled: StrictBool
    side_effects_stated: StrictBool
    callback_reason: Literal["missing_information", "consultation_needed", "invalid_refill_amount", "invalid_dob", "none"]

# The minimal prompt: name the fields and their legal values; explain nothing (workflow A.3).
SYSTEM = ("Read the pharmacy refill request inside <message> and reply with one JSON object with exactly these keys: "
          "patient_name (string), date_of_birth (string), medication (string), refill_days (integer), "
          "is_controlled (true or false), side_effects_stated (true or false), "
          "callback_reason (one of: missing_information, consultation_needed, invalid_refill_amount, invalid_dob, none). "
          "Reply with only the JSON object.")

# provided: derived from your contract
FIELDS = list(Record.model_fields)
LEGAL = {f: list(a.__args__) for f, a in Record.__annotations__.items() if getattr(a, "__origin__", None) is Literal}
BOOL_FIELDS = [f for f, i in Record.model_fields.items() if i.annotation is bool]     # StrictBool fields
print("fields:", FIELDS); print("fixed-choice fields:", LEGAL); print("true/false fields:", BOOL_FIELDS)
````

## Task 2 — The data (15 points)

*Mirrors the course's own ticket generator* ([`make_support_tickets.py`](https://github.com/prof-tcsmith/6564F26-DATA/blob/main/week-06-prompting-apis-finetuning/data/make_support_tickets.py)) *and lecture: Opening, "The bar to clear" (the honesty check).*

You need a **training set** (at least 300 rows) and a **held-out test set** (at least 40 rows), each with a
`text` column and one column per contract field, as strings (`"true"`/`"false"` for booleans). Three ways to
get them, all acceptable:

- **Templates**: sentence templates with slots, filled at random, the way the course's generator works.
  Write the test set from templates the training set never uses.
- **An AI assistant**: ask it for labelled examples in bulk (log the transcript), then verify a sample by
  hand. Keep the test set's phrasings distinct from training.
- **A public dataset** you relabel to your contract, fetched by URL inside the notebook.

Whatever the route, `make_data()` must **build or download it here**, deterministically, so the grader
can re-run the notebook. Then verify at least ten labels by hand and say so in the answer.


````python
# --- 2a: make_data() -> (train_df, test_df) ---------------------------------------------------
# Both DataFrames: a "text" column plus one column per field in FIELDS, all values as strings.
# Deterministic (use random.Random(SEED)); no phrasing in test_df may appear in train_df.
def make_data() -> tuple[pd.DataFrame, pd.DataFrame]:
    rng = random.Random(SEED)                               # one seeded generator: the same rows on every run
    MEDS = [("lisinopril", False), ("metformin", False), ("atorvastatin", False), ("sertraline", False),
            ("levothyroxine", False), ("alprazolam", True), ("oxycodone", True), ("tramadol", True), ("zolpidem", True)]   # (name, controlled?)
    FIRST = ["Maria", "James", "Aisha", "Robert", "Linda", "Kenji", "Priya", "Thomas", "Carmen", "Derek", "Olga", "Samuel"]
    LAST = ["Alvarez", "Bennett", "Chen", "Dubois", "Eze", "Fischer", "Gupta", "Hughes", "Ibarra", "Johansson", "Kowalski", "Lopez"]

    # Two phrase families with no shared wording: training rows use TRAIN, held-out test rows use TEST.
    TRAIN = {"open": ["Refill request, typed from the fax:", "Good morning, sending a refill request.", "Refill request by email:"],
             "name": "Patient name is {name}.", "no_name": "The patient name box was left blank.",
             "dob": "Date of birth {dob}.", "med": "Please refill {med}, a {days}-day supply.",
             "no_med": "Please refill the medication, a {days}-day supply, name not written.",
             "side": "They have had side effects since the last fill: dizziness and nausea.",
             "no_side": "No side effects reported.", "close": ["Thank you.", "Thanks in advance.", "Thank you for your help."]}
    TEST = {"open": ["Fax received, I need a renewal.", "Email from a patient asking for a refill:", "Hello, my prescription has run out."],
            "name": "Name on the form: {name}.", "no_name": "No name appears on the form.",
            "dob": "Born {dob}.", "med": "The medicine is {med} and I am out, need {days} days.",
            "no_med": "Medicine not filled in, I need {days} days of it.",
            "side": "Since my last fill I keep getting dizzy and nauseous.",
            "no_side": "Nothing unusual with the last fill.", "close": ["Regards.", "Cheers, thanks.", "Take care."]}

    def make_row(reason: str, ph: dict) -> dict:
        name = f"{rng.choice(FIRST)} {rng.choice(LAST)}"
        med, controlled = rng.choice(MEDS)
        month, day, year = rng.randint(1, 12), rng.randint(1, 28), rng.randint(1940, 2005)
        dob = f"{month:02d}/{day:02d}/{year}"
        days = rng.randint(7, 30) if controlled else rng.randint(7, 90)
        side = reason == "consultation_needed"
        if reason == "missing_information":                 # blank the name, or the medication
            if rng.random() < 0.5:
                name = ""
            else:
                med, controlled = "", False
        elif reason == "invalid_dob":                       # month 13, or February 30
            dob = rng.choice([f"13/{day:02d}/{year}", f"02/30/{year}"])
        elif reason == "invalid_refill_amount":             # over 90 days, or over 30 for a controlled drug
            days = rng.randint(31, 180) if controlled else rng.randint(91, 365)
        parts = [rng.choice(ph["open"]),
                 ph["name"].format(name=name) if name else ph["no_name"],
                 ph["dob"].format(dob=dob),
                 ph["med"].format(med=med, days=days) if med else ph["no_med"].format(days=days),
                 ph["side"] if side else (ph["no_side"] if rng.random() < 0.5 else ""),
                 rng.choice(ph["close"])]
        return {"text": " ".join(p for p in parts if p), "patient_name": name, "date_of_birth": dob,
                "medication": med, "refill_days": str(days), "is_controlled": str(controlled).lower(),
                "side_effects_stated": str(side).lower(), "callback_reason": reason}

    train_rows = [make_row(r, TRAIN) for r in LEGAL["callback_reason"] for _ in range(80)]   # 80 per label
    rng.shuffle(train_rows)
    test_rows = [make_row(r, TEST) for r in LEGAL["callback_reason"] for _ in range(12)]    # 12 per label
    return pd.DataFrame(train_rows), pd.DataFrame(test_rows)

_data = attempt("Task 2", make_data)
if _data is not None:
    train, test = _data
    train, test = train.astype(str), test.astype(str)
    print(f"train {len(train)} rows · test {len(test)} rows · columns {list(train.columns)}")
    display(train.sample(3, random_state=SEED))
````

````python
# --- self-check, Task 2 (provided) ---------------------------------------------------------------
def _check_task2():
    if need(_data): return None
    ok, names = [], []
    def add(name, cond): names.append(name); ok.append(bool(cond))
    add("train has 300+ rows", len(train) >= 300); add("test has 40+ rows", len(test) >= 40)
    add("columns = text + every field", set(train.columns) == {"text", *FIELDS} == set(test.columns))
    for f, legal in LEGAL.items():
        add(f"{f}: only legal values", set(train[f]) | set(test[f]) <= set(legal))
        add(f"{f}: every value appears 10+ times in train", train[f].value_counts().min() >= 10 and set(train[f]) == set(legal))
    def as_obj(r): return {f: (r[f] == "true") if f in BOOL_FIELDS else r[f] for f in FIELDS}
    try:
        for _, r in pd.concat([train, test]).iterrows(): Record.model_validate_json(json.dumps(as_obj(r)))
        add("every row validates against Record", True)
    except ValidationError as e:
        add(f"every row validates against Record ({e.errors()[0]['loc']}: {e.errors()[0]['msg']})", False)
    add("no test text appears in train", len(set(test.text) & set(train.text)) == 0)
    # phrasing check: a test text that shares 8 consecutive words with a training text is a reused template
    grams = lambda s, n=8: {" ".join(w[i:i + n]) for w in [s.lower().split()] for i in range(max(1, len(w) - n + 1))}
    train_grams = set().union(*(grams(t) for t in train.text))
    reused = sum(bool(grams(t) & train_grams) for t in test.text)
    add(f"test phrasings unseen in training (8-word overlap: {reused} of {len(test)}; must be < 20%)", reused < 0.2 * len(test))
    for n, o in zip(names, ok): print(f"  {'PASS' if o else 'FAIL'}  {n}")
    return all(ok)

T2_OK = _check_task2()
````

All 3 models are licenced under ollama (apache-2.0) but we will chose Qwen/Qwen2.5-0.5B-Instruct, because the breif lays out that what the model is doing is extracting text from a form, and then making judgement on what the callback_reason field should be.
````python
# --- the model, the helpers and the scorer (provided; workflow A.3) -------------------------------
from transformers import AutoModelForCausalLM, AutoTokenizer

if T1_OK and T2_OK and "MODEL_ID" in globals():
    tok = AutoTokenizer.from_pretrained(MODEL_ID); tok.padding_side = "left"
    model = AutoModelForCausalLM.from_pretrained(MODEL_ID).to(dtype=torch.float32).to(device)
    N_PARAMS = sum(p.numel() for p in model.parameters())
    print(f"{MODEL_ID}: {N_PARAMS/1e6:.0f} M numbers · window {model.config.max_position_embeddings:,} tokens")
else:
    print("SKIPPED  finish Tasks 1-3a first.")

def messages(text): return [{"role": "system", "content": SYSTEM}, {"role": "user", "content": f"<message>\n{text}\n</message>"}]
def chat(m): return tok.apply_chat_template(m, tokenize=False, add_generation_prompt=True)

@torch.no_grad()
def generate(m, texts, batch=8, max_new=80):
    out = []
    for i in range(0, len(texts), batch):
        enc = tok([chat(messages(t)) for t in texts[i:i + batch]], return_tensors="pt", padding=True).to(device)
        g = m.generate(**enc, max_new_tokens=max_new, do_sample=False, temperature=None, top_p=None, top_k=None, pad_token_id=tok.pad_token_id)
        out += [tok.decode(x[enc["input_ids"].shape[1]:], skip_special_tokens=True).strip() for x in g]
    return out

def check(raw):
    """The hard guardrail: a Record, or None. Never raises."""
    m = re.search(r"\{.*?\}", raw, re.S)
    if not m: return None
    try: return Record.model_validate_json(m.group(0))
    except ValidationError: return None

def score(raws, gold_df, name):
    """valid, each field, and ALL fields right, on gold_df. Field values compared as lower-case strings."""
    objs = [check(r) for r in raws]
    right = {f: 0 for f in FIELDS}; exact = 0
    for o, (_, g) in zip(objs, gold_df.iterrows()):
        if o is None: continue
        d = {k: str(v).lower() for k, v in o.model_dump().items()}
        ok = [d[f] == str(g[f]).lower() for f in FIELDS]
        for f, k in zip(FIELDS, ok): right[f] += k
        exact += all(ok)
    valid = sum(o is not None for o in objs)
    print(f"{name:<30} valid {valid:>3}/{len(gold_df)}  " + "  ".join(f"{f} {right[f]:>3}" for f in FIELDS) + f"   ALL RIGHT {exact:>3}/{len(gold_df)}")
    return {"name": name, "valid": valid, **right, "exact": exact}

RESULTS = []
````

## Task 4 — Fine-tune with LoRA (25 points)

*Mirrors workflow A.4–A.6. Lecture: Block 5, "LoRA: freeze W, learn two thin tables", "The data is the product", "The honest evaluation".*

Two functions are yours: `gold_json`, the answer the model is taught for a row, and `build_example`, which
turns a row into token ids and labels with the **prompt masked out** (`-100`) so only the answer is graded.
Then choose the three LoRA settings and justify them in one line each. The loop, the evaluation and the
save are provided.


````python
# --- 4a: the answer for one row, as the JSON text the model should write ----------------------
# Booleans must be real JSON true/false, not the strings "true"/"false" your DataFrame holds (BOOL_FIELDS helps).
def gold_json(row) -> str:
    obj = {}                                                    # the answer, built one field at a time
    for f in FIELDS:
        v = getattr(row, f)                                     # this row's value for field f, e.g. row.refill_days
        if f in BOOL_FIELDS:
            v = str(v).lower() == "true"                        # "true" -> True, "false" -> False
        elif Record.model_fields[f].annotation is int:
            v = int(v)                                          # "14" -> 14, so the JSON holds a number
        obj[f] = v
    return json.dumps(obj)                                      # dict -> JSON text, keys in FIELDS order

# --- 4b: one training example: (input_ids, labels) with the prompt marked -100 -----------------
def build_example(row, maxlen: int = 288) -> tuple[list[int], list[int]]:
    prompt_ids = tok(chat(messages(row.text)), add_special_tokens=False)["input_ids"]          # what the model reads
    answer_ids = tok(gold_json(row) + tok.eos_token, add_special_tokens=False)["input_ids"]    # what it must write, then <|im_end|>
    ids = prompt_ids + answer_ids                               # one sequence: prompt, then answer
    if len(ids) > maxlen: raise ValueError(f"example is {len(ids)} tokens, longer than maxlen={maxlen}")
    labels = [-100] * len(prompt_ids) + answer_ids              # -100 = "do not grade"; answer tokens keep their ids
    return ids, labels

# --- 4c: LoRA settings, one line of justification each in the answer below ---------------------
LORA_R = 16                 # rank: 16 small trainable tables per layer, ample for one short schema and 400 examples
LORA_ALPHA = 32             # scale: alpha / r = 2 lets the adapter's update override the base model's habits
TARGETS = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]   # every linear layer in each block
````

````python
# --- self-check, Task 4 (provided) ---------------------------------------------------------------
def _check_task4():
    if need(_data) or "model" not in globals(): return None
    r0 = next(train.itertuples())          # rows arrive as namedtuples, as in the training loop
    g = attempt("gold_json", gold_json, r0)
    if g is None: return None
    ok, names = [], []
    def add(n, c): names.append(n); ok.append(bool(c))
    try:
        obj = Record.model_validate_json(g); add("gold_json validates against Record", True)
        add("gold_json matches the row", all(str(v).lower() == str(getattr(r0, k)).lower() for k, v in obj.model_dump().items()))
    except ValidationError as e:
        add(f"gold_json validates against Record ({e.errors()[0]['msg']})", False)
    ex = attempt("build_example", build_example, r0)
    if ex is None: return None
    ids, lab = ex
    add("ids and labels have the same length", len(ids) == len(lab))
    n_prompt = len(tok(chat(messages(r0.text)), add_special_tokens=False)["input_ids"])
    add("the whole prompt is masked with -100", lab[:n_prompt] == [-100] * n_prompt)
    add("the answer is NOT masked", any(x != -100 for x in lab[n_prompt:]) and lab[n_prompt:] == ids[n_prompt:])
    add("the answer ends with the end-of-turn token", ids[-1] == tok.convert_tokens_to_ids(tok.eos_token) or ids[-1] == tok.eos_token_id)
    add("LoRA settings set", all(v is not None for v in (LORA_R, LORA_ALPHA, TARGETS)) and 4 <= LORA_R <= 64)
    for n, o in zip(names, ok): print(f"  {'PASS' if o else 'FAIL'}  {n}")
    return all(ok)

T4_OK = _check_task4()
````

````python
# --- train (provided; workflow A.4). Two to three minutes on a T4. --------------------------------
import subprocess, sys
# Uninstall incompatible torchao version that causes peft initialization crash
subprocess.run([sys.executable, "-m", "pip", "uninstall", "-y", "torchao"], capture_output=True)

from peft import LoraConfig, get_peft_model, PeftModel

if T4_OK:
    if isinstance(model, PeftModel):
        model = model.unload()
    cfg = LoraConfig(r=LORA_R, lora_alpha=LORA_ALPHA, lora_dropout=0.05, bias="none", task_type="CAUSAL_LM", target_modules=TARGETS)
    peft_model = get_peft_model(model, cfg)
    trainable = sum(p.numel() for p in peft_model.parameters() if p.requires_grad)
    print(f"trainable: {trainable/1e6:.1f} M of {N_PARAMS/1e6:.0f} M numbers ({100*trainable/N_PARAMS:.2f}%)")
    examples = [build_example(r) for r in train.itertuples()]
    order = list(range(len(examples))); random.Random(SEED).shuffle(order)
    opt = torch.optim.AdamW([p for p in peft_model.parameters() if p.requires_grad], lr=2e-4)
    peft_model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
    peft_model.enable_input_require_grads(); model.config.use_cache = False
    peft_model.train(); t0 = time.time(); losses = []
    for step, i in enumerate(range(0, len(order), 8)):
        batch = [examples[j] for j in order[i:i + 8]]
        L = max(len(a) for a, _ in batch)
        ids = torch.tensor([a + [tok.pad_token_id] * (L - len(a)) for a, _ in batch]).to(device)
        lab = torch.tensor([b + [-100] * (L - len(b)) for _, b in batch]).to(device)
        att = torch.tensor([[1] * len(a) + [0] * (L - len(a)) for a, _ in batch]).to(device)
        loss = peft_model(input_ids=ids, attention_mask=att, labels=lab).loss
        loss.backward(); opt.step(); opt.zero_grad(set_to_none=True); losses.append(loss.item())
        if step % 15 == 0: print(f"step {step:>3}  loss {loss.item():.3f}")
    peft_model.gradient_checkpointing_disable(); model.config.use_cache = True; peft_model.eval()
    train_secs = time.time() - t0
    print(f"{len(losses)} steps in {train_secs:.0f} s on {device}; loss {losses[0]:.2f} -> {np.mean(losses[-5:]):.3f}")
else:
    peft_model = None; print("SKIPPED  Task 4's self-check has not passed.")
````

````python
# --- the honest evaluation (provided; workflow A.5): same test set, same scorer, adapter off vs on --
if peft_model is not None:
    tuned_raw = generate(peft_model, list(test.text))
    RESULTS.append(score(tuned_raw, test, "fine-tuned + minimal prompt"))
    with peft_model.disable_adapter():                     # the controlled comparison: only the adapter differs
        off_raw = generate(peft_model, list(test.text))
    RESULTS.append(score(off_raw, test, "same model, adapter OFF"))
    display(pd.DataFrame(RESULTS).set_index("name"))
    weakest = min(FIELDS, key=lambda f: RESULTS[-2][f])
    print(f"weakest field after tuning: {weakest} ({RESULTS[-2][weakest]} of {len(test)})")
    for t, r in list(zip(test.text, tuned_raw))[:3]:
        print(f"\n{t[:110]}\n -> {r[:130]}")
else:
    print("SKIPPED")


#Added for more context
pred = [(check(r).callback_reason if check(r) else "INVALID") for r in tuned_raw]
display(pd.crosstab(test.callback_reason.values, pd.Series(pred, name="predicted"), rownames=["actual"]))

bad = test[[p != a for p, a in zip(pred, test.callback_reason)]]
display(bad[["text", "patient_name", "date_of_birth", "medication", "refill_days", "callback_reason"]])
````
