````python
# --- the model, the helpers and the scorer (provided; workflow A.3) -------------------------------
from transformers import AutoModelForCausalLM, AutoTokenizer

if T1_OK and T2_OK and "MODEL_ID" in globals():
    tok = AutoTokenizer.from_pretrained(MODEL_ID); tok.padding_side = "left"
    model = AutoModelForCausalLM.from_pretrained(MODEL_ID).to(dtype=torch.float32).to(device)
    N_PARAMS = sum(p.numel() for p in model.parameters())
    print(f"{MODEL_ID}: {N_PARAMS/1e6:.0f} M numbers · window {model.config.max_position_embeddings:,} tokens")
else:
    print("SKIPPED  finish Tasks 1-3a first.")

def messages(text): return [{"role": "system", "content": SYSTEM}, {"role": "user", "content": f"<message>\n{text}\n</message>"}]
def chat(m): return tok.apply_chat_template(m, tokenize=False, add_generation_prompt=True)

@torch.no_grad()
def generate(m, texts, batch=8, max_new=80):
    out = []
    for i in range(0, len(texts), batch):
        enc = tok([chat(messages(t)) for t in texts[i:i + batch]], return_tensors="pt", padding=True).to(device)
        g = m.generate(**enc, max_new_tokens=max_new, do_sample=False, temperature=None, top_p=None, top_k=None, pad_token_id=tok.pad_token_id)
        out += [tok.decode(x[enc["input_ids"].shape[1]:], skip_special_tokens=True).strip() for x in g]
    return out

def check(raw):
    """The hard guardrail: a Record, or None. Never raises."""
    m = re.search(r"\{.*?\}", raw, re.S)
    if not m: return None
    try: return Record.model_validate_json(m.group(0))
    except ValidationError: return None

def score(raws, gold_df, name):
    """valid, each field, and ALL fields right, on gold_df. Field values compared as lower-case strings."""
    objs = [check(r) for r in raws]
    right = {f: 0 for f in FIELDS}; exact = 0
    for o, (_, g) in zip(objs, gold_df.iterrows()):
        if o is None: continue
        d = {k: str(v).lower() for k, v in o.model_dump().items()}
        ok = [d[f] == str(g[f]).lower() for f in FIELDS]
        for f, k in zip(FIELDS, ok): right[f] += k
        exact += all(ok)
    valid = sum(o is not None for o in objs)
    print(f"{name:<30} valid {valid:>3}/{len(gold_df)}  " + "  ".join(f"{f} {right[f]:>3}" for f in FIELDS) + f"   ALL RIGHT {exact:>3}/{len(gold_df)}")
    return {"name": name, "valid": valid, **right, "exact": exact}

RESULTS = []
````
